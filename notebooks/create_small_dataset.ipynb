```python
import json

with open("../data/medical-diff-vqa-1.0.1/all_diseases.json") as f:
    data = json.load(f)  # this is a list of dicts

# filter elements where subject_id starts with '10'
filtered_data = [item for item in data if str(item["subject_id"]).startswith("10000935")]

# save filtered results back to JSON
with open(
    f"../data/medical-diff-vqa-1.0.1/all_diseases_p{filter_string}.json", "w"
) as f:
    json.dump(filtered_data, f, indent=4)

print(f"Saved {len(filtered_data)} records to filtered_subjects.json")
```
